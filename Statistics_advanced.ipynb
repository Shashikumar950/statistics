 1 What is a random variable in probability theory?

    ->A random variable is a function that assigns numerical values to the outcomes of a probabilistic experiment. In simpler terms, it’s a way to translate uncertain events into numbers so we can analyze them mathematically.

    Definition
   A random variable is a variable whose value depends on the outcome of a random experiment.
   Formally, it is a function from the sample space (all possible outcomes) to the set of real numbers.
  The actual number we get after the experiment is called a realization of the random variable.

    Types of Random Variables
   Discrete random variable: Takes on a finite or countable set of values (e.g., number of heads in 10 coin flips).
   Continuous random variable: Can take any value within an interval of real numbers (e.g., the exact height of a randomly chosen person).

    Why It Matters
     Random variables are the foundation of probability and statistics because they:
     Allow us to quantify uncertainty.
     Enable the use of probability distributions (like binomial, normal, Poisson).
     Provide tools for statistical inference (confidence intervals, hypothesis testing, etc.).


    Example
     Tossing a fair coin:
     Sample space: {Heads, Tails}
     Define random variable
          X = 1 if Heads
          X = 0 if Tails
         Here, X is a discrete random variable.

    Measuring rainfall in a day:
     Random variable Y = amount of rainfall (in mm).
     Y can take any non-negative real value, so it’s a continuous random variable.

  6: What is the Central Limit Theorem (CLT), and why is it critical in statistics?

    ->central Limit Theorem (CLT)
    The Central Limit Theorem (CLT) states that the sampling distribution of the sample mean will approach a normal (Gaussian) distribution as the sample size becomes large, regardless of the shape of the original population distribution, provided the population has a finite mean and variance.


    Formally:
     If you take many random samples of size n from any population,
     Then the distribution of their sample means tends toward a normal distribution as n increases,
     With mean 𝜇(the population mean) and standard deviation 𝜎/𝑛

     Why CLT Is Critical in Statistics
     Enables Normal Approximation
     Even if the original data are skewed or non-normal, CLT allows statisticians to use normal distribution methods for large samples.

     Foundation of Statistical Inference
     Many core techniques—confidence intervals, hypothesis tests, t-tests, and z-tests—depend on the CLT.

     Predicts Sampling Behavior
     It explains why sample means are more stable and less variable than individual observations.

     Works Across Fields
     CLT is essential in economics, science, engineering, medicine, and social sciences for analyzing real-world data.


  7: What is the significance of confidence intervals in statistical analysis?

    -> Confidence intervals are fundamental in statistical analysis because they provide more information than a single point estimate. Instead of just reporting a mean, proportion, or other statistic, confidence intervals give a range of plausible values for the population parameter, along with a measure of certainty.

      Significance of Confidence Intervals
     Precision of Estimates
      They show how reliable your sample estimate is. A narrow interval means high precision, while a wide interval indicates more uncertainty.

     Acknowledges Sampling Variability
     Different samples produce different results. Confidence intervals account for this variability by giving a range that likely contains the true population value.

     Supports Decision-Making
      In research or business, confidence intervals help determine whether results are meaningful. For example, if a confidence interval for a treatment effect does not include zero, it suggests the treatment is effective.

        Linked to Hypothesis Testing
     A 95% confidence interval that excludes the null hypothesis value (like zero difference) implies statistical significance at the 5% level.

      Improves Communication
     Saying “the average sales are between 240 and 260 units with 95% confidence” is more informative than just “the average is 250.”

  9 Write a Python program to generate 1000 random numbers from a normal
   distribution with mean = 50 and standard deviation = 5. Compute its mean and standard deviation using numpy ,and draw a histogram to visualize the distribution.
   
  


```python
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Generate 1000 random numbers from a normal distribution
mean = 50
std_dev = 5
data = np.random.normal(mean, std_dev, 1000)

# Step 2: Compute mean and standard deviation using NumPy
calculated_mean = np.mean(data)
calculated_std = np.std(data)

print("Calculated Mean:", calculated_mean)
print("Calculated Standard Deviation:", calculated_std)

# Step 3: Draw a histogram to visualize the distribution
plt.hist(data, bins=30, color='skyblue', edgecolor='black')
plt.title("Histogram of Normal Distribution (mean=50, std=5)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

  10: You are working as a data analyst for a retail company. The company has collected daily sales data for 2 years and wants you to identify the overall sales trend .
   daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
                         235, 260, 245, 250, 225, 270, 265, 255, 250, 260]
   ● Explain how you would apply the Central Limit Theorem to estimate the average saleswith a 95% confidence interval .
   ● Write the Python code to compute the mean sales and its confidence interval.

    ->Central Limit Theorem (CLT):  
    The CLT states that if we take repeated random samples from a population, the distribution of the sample means will approximate a normal distribution as the sample size grows, regardless of the population’s distribution.

     Application here:
      We have daily sales data. To estimate the average daily sales, we can treat this dataset as a sample from the larger population (all possible daily sales).
      By CLT, the sample mean will follow a normal distribution if the sample size is reasonably large.
     This allows us to construct a confidence interval (CI) for the mean using the formula:
                                  s
                   cl = x + Za/z.---
                                  n
      𝑥= sample mean
      𝑠= sample standard deviation
      𝑛= sample size
      𝑍
      𝛼/2= critical value from the standard normal distribution (for 95% CI, cl, Z = 1.96)

     Interpretation:  
     The 95% CI gives a range where we are 95% confident the true average daily sales lie.


```python
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt

# Daily sales data
daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
               235, 260, 245, 250, 225, 270, 265, 255, 250, 260]

# Convert to numpy array
data = np.array(daily_sales)

# Compute mean and standard deviation
mean_sales = np.mean(data)
std_sales = np.std(data, ddof=1)  # sample standard deviation
n = len(data)

# 95% confidence interval using CLT
confidence_interval = st.norm.interval(
    0.95, loc=mean_sales, scale=std_sales/np.sqrt(n)
)

print("Mean Sales:", mean_sales)
print("95% Confidence Interval:", confidence_interval)

# Plot histogram
plt.hist(data, bins=8, color='skyblue', edgecolor='black')
plt.axvline(mean_sales, color='red', linestyle='dashed', linewidth=2, label=f'Mean = {mean_sales:.2f}')
plt.legend()
plt.title("Daily Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```
